## Tugas 1
Analisa korelasi antara `temperature_fahrenheit`, `pressure_mb`, dan `air_quality_Sulphur_dioxide` (dataset `GlobalWeatherRepository.csv`)

```python
import pandas as pd
path = "GlobalWeatherRepository.csv"
df = pd.read_csv(path)
df
```

#### Penjelasan:
Kode di atas hanya untuk buka file CSV-nya menggunakan pandas, kemudian akan disimpan ke variabel `df` agar bisa diolah. Baris terakhir (`df`) itu untuk nampilin isinya, sekalian ngecek datanya kebaca dengan benar atau tidak.

```python
korelasi = df[['temperature_fahrenheit', 'pressure_mb', 'air_quality_Sulphur_dioxide']].corr()
korelasi
```

#### Penjelasan:
Kode di atas memilih 3 kolom yaitu `temperature_fahrenheit`, `pressure_mb`, dan `air_quality_Sulphur_dioxide` dari DataFrame `df`, kemudian akan dicek korelasinya pakai `.corr()`. Fungsi ini akan memberi tahu seberapa nyambung satu variabel sama variabel lainnya.

Dari hasil di atas terlihat bahwa `temperature_fahrenheit` vs `pressure_mb` bernilai -0.486833 (korelasi negatif sedang, artinya semakin tinggi suhu maka tekanan udara cenderung menurun), sedangkan korelasi `air_quality_Sulphur_dioxide` terhadap dua variabel lainnya sangat lemah (mendekati 0), menandakan hampir tidak ada hubungan.
