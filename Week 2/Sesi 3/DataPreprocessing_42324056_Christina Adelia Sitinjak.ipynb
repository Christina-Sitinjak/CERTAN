## Outliers

```python
path = "weight-height_outlier.csv"
df2 = pd.read_csv(path)
df2
```

```python
import seaborn as sns
sns.boxplot(df2['Height'])
```

```python
import numpy as np
# Identifikasi Q1 Q3
q1, q3 = np.percentile(df2['Height'], [25, 75])
# Hitung IQR
iqr = q3 - q1
# Cari lower dan upper bounds
lower_bound = q1 - (1.5 * iqr)
upper_bound = q3 + (1.5 * iqr)
# Hapus outliers
clean_data = df2[(df2['Height'] >= lower_bound)
                  & (df2['Height'] <= upper_bound)]

sns.boxplot(clean_data['Height'])
```

```python
sns.boxplot(df2['Weight'])
```

```python
# Identifikasi Q1 Q3
q1, q3 = np.percentile(df2['Weight'], [25, 75])
# Hitung IQR
iqr = q3 - q1
# Cari lower dan upper bounds
lower_bound = q1 - (1.5 * iqr)
upper_bound = q3 + (1.5 * iqr)
# Hapus outliers
clean_data = df2[(df2['Weight'] >= lower_bound)
                  & (df2['Weight'] <= upper_bound)]

sns.boxplot(clean_data['Weight'])
```

```python
import matplotlib.pyplot as plt

# Membuat daftar kolom yang ingin dianalisis, yaitu "Weight" dan "Height".
columns_to_plot = ['Weight', 'Height']

# Memastikan kolom ada di dataset
columns_to_plot = [col for col in columns_to_plot if col in df2.columns]

fig, axs = plt.subplots(len(columns_to_plot), 1, dpi=95, figsize=(7, 5 * len(columns_to_plot)))

if len(columns_to_plot) == 1:
    axs = [axs]

for i, col in enumerate(columns_to_plot):
    axs[i].boxplot(df2[col], vert=False)
    axs[i].set_ylabel(col)

plt.show()
```

```python
def remove_outliers_sequentially(df2, columns):
    df_cleaned = df2.copy()
    for col in columns:
        q1, q3 = np.percentile(df_cleaned[col], [25, 75])
        iqr = q3 - q1
        lower_bound = q1 - (1.5 * iqr)
        upper_bound = q3 + (1.5 * iqr)
        df_cleaned = df_cleaned[(df_cleaned[col] >= lower_bound) & (df_cleaned[col] <= upper_bound)]
    return df_cleaned

# Terapkan fungsi untuk menghapus outliers dari "Height" dan "Weight"
clean_data = remove_outliers_sequentially(df2, ['Height', 'Weight'])

# Plot boxplots setelah menghapus outliers
columns_to_plot = ['Weight', 'Height']
fig, axs = plt.subplots(len(columns_to_plot), 1, dpi=95, figsize=(7, 5 * len(columns_to_plot)))

if len(columns_to_plot) == 1:
    axs = [axs]

for i, col in enumerate(columns_to_plot):
    axs[i].boxplot(clean_data[col], vert=False)
    axs[i].set_ylabel(col)

plt.show()
```

```python
from sklearn.preprocessing import MinMaxScaler
scaler = MinMaxScaler(feature_range=(0, 1))

# Memisahkan kolom numerik dan non-numerik
numeric_cols = clean_data.select_dtypes(include=[np.number]).columns  # Ambil kolom numerik
non_numeric_cols = clean_data.select_dtypes(exclude=[np.number]).columns  # Ambil kolom non-numerik

rescaled_numeric = scaler.fit_transform(clean_data[numeric_cols])
rescaled_numeric[:]
```

```python
# Konversi kembali hasil scaling ke DataFrame
rescaled_df = pd.DataFrame(rescaled_numeric, columns=numeric_cols, index=clean_data.index)

# Gabungkan kembali dengan kolom non-numerik tanpa mengubah jumlah baris
final_df = pd.concat([clean_data[non_numeric_cols], rescaled_df], axis=1)

final_df
```

## Tugas 2 - Handling Missing Value (Titanic-Dataset.csv)

```python
import pandas as pd
import numpy as np

path = "Titanic-Dataset.csv"
df_titanic = pd.read_csv(path)
df_titanic
```

#### Penjelasan:
Kode di atas untuk membuka file Titanic-nya mnenggunakan pandas, lalu disimpan ke `df_titanic`.

```python
df_titanic.isnull().sum()
```

#### Penjelasan:
Ini dibuat untuk ngecek kolom mana aja yang ada data kosongnya. Ternyata ada 3 kolom bermasalah:
- `Age` kosong di 177 baris
- `Cabin` kosong di 687 baris (paling parah, dari 891 total data)
- `Embarked` cuma kosong di 2 baris

```python
# Age -> isi dengan median (177 data hilang, numerik)
df_titanic['Age'] = df_titanic['Age'].fillna(df_titanic['Age'].median())

df_titanic['Age'].isnull().sum()
```

#### Penjelasan:
Untuk `Age`, saya isi pakai median (bukan mean), karena umur biasanya (ada beberapa penumpang yang jauh lebih tua dari yang lain), jadi median lebih aman daripada mean yang mudah ketarik dengan nilai ekstrem.

```python
# Cabin -> hapus kolomnya, karena 687 dari 891 data (77%) kosong
df_titanic = df_titanic.drop(columns='Cabin')

df_titanic.columns
```

#### Penjelasan:
Kolom `Cabin` ini sangat banyak yg kosong, 77% datanya hilang. Kalau dipaksa diisi (misal pakai modus), hasilnya malah jadi nggak representatif dan bisa bikin analisa makin salah. Makanya kolom ini saya hapus aja sekalian.

```python
# Embarked -> isi dengan modus, karena hanya 2 data yang kosong
df_titanic['Embarked'] = df_titanic['Embarked'].fillna(df_titanic['Embarked'].mode().iloc[0])

df_titanic['Embarked'].isnull().sum()
```

#### Penjelasan:
Karena `Embarked` hanya kosong di 2 baris, saya isi pakai modus (nilai yang paling sering muncul). Ini aman karena dampaknya ke keseluruhan data sangat kecil.

```python
df_titanic.isnull().sum()
```

#### Penjelasan:
Ini pengecekan terakhir untuk memastikan semua missing value. Jika hasilnya semua kolom nilainya 0, maka tidak ada data yang kosong lagi.

## Tugas 3 - Handling Outliers (diabetes.csv)
Kolom yang dipilih untuk dideteksi: **BloodPressure**, **Insulin**, **BMI**.

```python
path = "diabetes.csv"
df_diabetes = pd.read_csv(path)
df_diabetes
```

#### Penjelasan:
Sama seperti sebelumnya, untuk membuka file `diabetes.csv`-nya dan disimpan ke `df_diabetes`.

```python
import matplotlib.pyplot as plt

kolom_dipilih = ['BloodPressure', 'Insulin', 'BMI']

fig, axs = plt.subplots(len(kolom_dipilih), 1, dpi=95, figsize=(7, 5 * len(kolom_dipilih)))

for i, col in enumerate(kolom_dipilih):
    axs[i].boxplot(df_diabetes[col], vert=False)
    axs[i].set_ylabel(col)

plt.suptitle('Sebelum Outlier Dihapus')
plt.show()
```

#### Penjelasan:
Saya memilih 3 kolom ini (`BloodPressure`, `Insulin`, `BMI`) karena pas dicek, ketiganya punya cukup banyak titik di luar whisker boxplot-nya, jadi cocok untuk dijadikan contoh penanganan outlier. Titik-titik bulat di luar 'kumis' boxplot itu yang disebut outlier.

```python
def remove_outliers_sequentially(data, columns):
    df_cleaned = data.copy()
    for col in columns:
        q1, q3 = np.percentile(df_cleaned[col], [25, 75])
        iqr = q3 - q1
        lower_bound = q1 - (1.5 * iqr)
        upper_bound = q3 + (1.5 * iqr)
        df_cleaned = df_cleaned[(df_cleaned[col] >= lower_bound) & (df_cleaned[col] <= upper_bound)]
    return df_cleaned

print('Jumlah baris sebelum hapus outlier:', df_diabetes.shape[0])

clean_diabetes = remove_outliers_sequentially(df_diabetes, kolom_dipilih)

print('Jumlah baris setelah hapus outlier:', clean_diabetes.shape[0])
```

#### Penjelasan:
Fungsi ini untuk menghitung batas bawah dan batas atas pakai metode IQR (Interquartile Range) buuntukat masing-masing kolom, terus buang baris yang nilainya di luar batas itu. Dilakuin bertahap (satu kolom demi satu kolom), jadi data yang tersisa itu telah 'bersih' dari outlier di ketiga kolom sekaligus.

Dari 768 baris data awal, setelah dibersihkan tersisa 685 baris. Artinya sekitar 83 baris data (10.8%) terdeteksi sebagai outlier di salah satu dari 3 kolom tersebut, dan sudah dibuang.

```python
fig, axs = plt.subplots(len(kolom_dipilih), 1, dpi=95, figsize=(7, 5 * len(kolom_dipilih)))

for i, col in enumerate(kolom_dipilih):
    axs[i].boxplot(clean_diabetes[col], vert=False)
    axs[i].set_ylabel(col)

plt.suptitle('Setelah Outlier Dihapus')
plt.show()
```

#### Penjelasan:
Ini dibuat untuk memastikan hasil pembersihan tadi beneran berhasil. Jika dibandingkan dengan boxplot yang pertama, sekarang titik-titik outlier di luar kumis boxplot telah jauh berkurang (atau bahkan hilang sama sekali) untuk ketiga kolom ini.
